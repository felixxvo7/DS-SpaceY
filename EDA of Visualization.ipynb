## Exploratory Data Analysis


```python
import pandas as pd
import requests
from io import BytesIO

# URL for the dataset
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"

# Fetch the data using requests
response = requests.get(URL)

# Load the content into a pandas DataFrame
dataset_part_2_csv = BytesIO(response.content)
df = pd.read_csv(dataset_part_2_csv)

# Display the first 5 rows of the DataFrame
df.head(5)
```

```python
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect = 5)
plt.xlabel("Flight Number",fontsize=20)
plt.ylabel("Pay load Mass (kg)",fontsize=20)
plt.show()
```

### TASK 1: Visualize the relationship between Flight Number and Launch Site

```python
# Scatter plot between Flight Number and Launch Site
plt.figure(figsize=(10, 6))
sns.scatterplot(x='FlightNumber', y='LaunchSite', data=df)
plt.title('Flight Number vs Launch Site')
plt.xlabel('Flight Number')
plt.ylabel('Launch Site')
plt.show()
```

### TASK 2: Visualize the relationship between Payload Mass and Launch Site


```python
# Scatter plot to visualize the relationship between Payload Mass and Launch Site with hue as Class
plt.figure(figsize=(12, 6))
sns.scatterplot(x='PayloadMass', y='LaunchSite', hue='Class', data=df, palette='viridis')

# Adding title and labels
plt.title('Payload Mass vs Launch Site with Class as Hue')
plt.xlabel('Payload Mass (kg)')
plt.ylabel('Launch Site')
plt.legend(title='Class')
plt.show()
```

### TASK  3: Visualize the relationship between success rate of each orbit type


```python
# Group the data by 'Orbits' and calculate the success rate (mean of 'Class')
orbit_success_rate = df.groupby('Orbit')['Class'].mean().reset_index()

# Create a bar chart to visualize the success rate for each orbit type
plt.figure(figsize=(12, 6))
sns.barplot(x='Orbit', y='Class', data=orbit_success_rate, palette='Blues')

# Adding title and labels
plt.title('Success Rate of Each Orbit Type')
plt.xlabel('Orbit Type')
plt.ylabel('Success Rate')
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.show()
```

### TASK  4: Visualize the relationship between FlightNumber and Orbit type


```python
# Scatter plot to visualize the relationship between FlightNumber and Orbit with hue as Class
plt.figure(figsize=(12, 6))
sns.scatterplot(x='FlightNumber', y='Orbit', hue='Class', data=df, palette='coolwarm', edgecolor=None)

# Adding title and labels
plt.title('Relationship Between Flight Number and Orbit Type with Class as Hue')
plt.xlabel('Flight Number')
plt.ylabel('Orbit Type')
plt.legend(title='Class')
plt.show()
```

### TASK  5: Visualize the relationship between Payload Mass and Orbit type


```python
# Scatter plot to visualize the relationship between Payload Mass and Orbit with hue as Class
plt.figure(figsize=(12, 6))
sns.scatterplot(x='PayloadMass', y='Orbit', hue='Class', data=df, palette='coolwarm', edgecolor=None)

# Adding title and labels
plt.title('Relationship Between Payload Mass and Orbit Type with Class as Hue')
plt.xlabel('Payload Mass (kg)')
plt.ylabel('Orbit Type')
plt.legend(title='Class')
plt.show()
```

### TASK  6: Visualize the launch success yearly trend


```python
# A function to Extract years from the date 
year=[]
def Extract_year():
    for i in df["Date"]:
        year.append(i.split("-")[0])
    return year
Extract_year()
df['Date'] = year
df.head()
    
```

```python

# Group by 'Year' and calculate the success rate (mean of 'Class')
yearly_success_rate = df.groupby('Date')['Class'].mean().reset_index()

# Plot a line chart for launch success rate over the years
plt.figure(figsize=(12, 6))
plt.plot(yearly_success_rate['Date'], yearly_success_rate['Class'], marker='o', color='b', linestyle='-', linewidth=2)

# Adding title and labels
plt.title('Launch Success Rate Trend Over the Years')
plt.xlabel('Date')
plt.ylabel('Success Rate')
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```
