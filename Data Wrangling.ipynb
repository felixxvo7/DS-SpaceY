### TASK 1: Calculate the number of launches on each site

The data contains several Space X  launch facilities: <a href='https://en.wikipedia.org/wiki/List_of_Cape_Canaveral_and_Merritt_Island_launch_sites'>Cape Canaveral Space</a> Launch Complex 40  <b>VAFB SLC 4E </b> , Vandenberg Air Force Base Space Launch Complex 4E <b>(SLC-4E)</b>, Kennedy Space Center Launch Complex 39A <b>KSC LC 39A </b>.The location of each Launch Is placed in the column <code>LaunchSite</code>


```python
# Apply value_counts() on the LaunchSite column
launch_counts = df['LaunchSite'].value_counts()

# Display the result
print(launch_counts)
```

### TASK 2: Calculate the number and occurrence of each orbit


```python
# Apply value_counts() on the Outcome column to determine the landing outcomes
landing_outcomes = df['Outcome'].value_counts()

# Display the result
print(landing_outcomes)
```

```python
# Print the index and the outcome for each landing outcome
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)

# Select bad outcomes by their index positions (e.g., 1, 3, 5, 6, 7)
bad_outcomes = set(landing_outcomes.index[[1, 3, 5, 6, 7]])

# Display the bad outcomes
print(bad_outcomes)
```

### TASK 4: Create a landing outcome label from Outcome column


```python
# Create the 'Class' column based on whether the outcome is a bad outcome (0 for bad, 1 for others)
df['Class'] = df['Outcome'].apply(lambda x: 0 if x in bad_outcomes else 1)

# Display the first 8 rows of the 'Class' column
print(df[['Class']].head(8))

# Display the first 5 rows of the DataFrame
print(df.head(5))

# Calculate the mean of the 'Class' column (proportion of good outcomes)
print(df["Class"].mean())
```

```python
# Count the number of launches from CCAFS SLC 40
launches_ccafs_slc40 = df[df['LaunchSite'] == 'CCAFS SLC 40'].shape[0]

# Display the result
print(launches_ccafs_slc40)
```

```python
# Count the occurrences of each orbit type in the Orbit column
orbit_counts = df['Orbit'].value_counts()

# Get the count for GTO (Geostationary Transfer Orbit)
gto_count = orbit_counts.get('GTO', 0)

# Display the result
print(gto_count)
# List of landing outcomes that represent a complete failure to land (you can adjust this based on your data)
failure_outcomes = ['Failure', 'Crash', 'Failed']

# Count the number of failures in the Outcome column
failure_count = df[df['Outcome'].isin(failure_outcomes)].shape[0]

# Display the result
print(failure_count)
```
